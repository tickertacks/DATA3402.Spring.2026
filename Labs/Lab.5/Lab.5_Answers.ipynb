1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.


```python
class matrix:
    # Initialization
    # A: Matrix(n, m)
    # B: Matrix([...], [...], [...], ...)
    # data will represent both n and the list
    def __init__(self, data, m=None):
        
        # Matrix(n, m)
        if isinstance(data, int) and isinstance(m, int):
            n = data
            self.rows = n
            self.cols = m
            
            # Zero matrix
            self.data = []
            for i in range(n):
                row = []
                for j in range(m):
                    row.append(0)
                self.data.append(row)
        
        # Matrix(list_of_lists)
        elif isinstance(data, list):
            
            # Checks for all rows as lists
            for row in data:
                if not isinstance(row, list):
                    raise TypeError("Matrix must be a list of lists")

            # Checks if all rows have the same length
            first_length = len(data[0])
            for row in data:
                if len(row) != first_length:
                    raise ValueError("Rows must have the same number of columns")

            self.rows = len(data)
            self.cols = first_length

            self.data = []
            for row in data:
                new_row = []
                for value in row:
                    new_row.append(value)
                self.data.append(new_row)

        else:
            raise ValueError("Invalid initialization")
                     
    # Indexing
    def __getitem__(self, key):

        # M[i, j] 
        if isinstance(key, tuple):
            i, j = key
            return self.data[i][j]

        # M[i][j]
        else:
            return self.data[key]

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.data[i][j] = value
        else:
            self.data[key] = value

    # Matrix assignment
    # Assigning values from another Matrix or a list of lists
    # A: If M_1 and M_2 are matrix instances M_1=M_2, set values of M_1 to those of M_2
    # B: Can assign from list of lists (e.g. M_2 being list of lists of correct size)

    def assign(self, other):
        # Matrices
        if isinstance(other, matrix):

            # Checks for same sized matrices
            if self.rows != other.rows or self.cols != other.cols:
                raise ValueError("Matrices must be the same size")

            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other.data[i][j]

        # List of lists
        elif isinstance(other, list):

            if len(other) != self.rows:
                raise ValueError("Incorrect number of rows")

            for row in other:
                if len(row) != self.cols:
                    raise ValueError("Incorrect number of columns")

            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other[i][j]

        else:
            raise TypeError("Invalid assignment")
            
    # Function for easy testing display
    def __repr__(self):
        s = ""
        for row in self.data:
            s += str(row) + "\n"
        return s
```

```python
class matrix:

    def __init__(self, data, m=None):

        if isinstance(data, int) and isinstance(m, int):
            n = data
            self.rows = n
            self.cols = m
            
            self.data = []
            for i in range(n):
                row = []
                for j in range(m):
                    row.append(0)
                self.data.append(row)
        
        elif isinstance(data, list):
            
            for row in data:
                if not isinstance(row, list):
                    raise TypeError("Matrix must be a list of lists")

            first_length = len(data[0])
            for row in data:
                if len(row) != first_length:
                    raise ValueError("Rows must have the same number of columns")

            self.rows = len(data)
            self.cols = first_length

            self.data = []
            for row in data:
                new_row = []
                for value in row:
                    new_row.append(value)
                self.data.append(new_row)

        else:
            raise ValueError("Invalid initialization")

# --------------------- New __getitem __ --------------------- # 
    def __getitem__(self, key):

        if isinstance(key, tuple):
            row_key, col_key = key
            
            # Convert single numbers to slices
            if isinstance(row_key, int):
                row_key = slice(row_key, row_key + 1)
                
            if isinstance(col_key, int):
                col_key = slice(col_key, col_key + 1)
                
            new_data = []
            # !!! An LLM helped me figure out the logic for this part:
            # .indices(length) adjusts slice
            # the * takes the tuple returned by .indices() and unpacks it into range
            # loop over selected rows
            for i in range(*row_key.indices(self.rows)):
                new_row = []
                # loop over selected columns
                for j in range(*col_key.indices(self.cols)):
                    new_row.append(self.data[i][j])
                new_data.append(new_row)

            return matrix(new_data)
        
        else:
            return self.data[key]
# ------------------------------------------------------------ #
        
    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.data[i][j] = value
        else:
            self.data[key] = value

    def assign(self, other):
        
        if isinstance(other, matrix):
            if self.rows != other.rows or self.cols != other.cols:
                raise ValueError("Matrices must be the same size")

            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other.data[i][j]

        elif isinstance(other, list):

            if len(other) != self.rows:
                raise ValueError("Incorrect number of rows")

            for row in other:
                if len(row) != self.cols:
                    raise ValueError("Incorrect number of columns")

            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other[i][j]

        else:
            raise TypeError("Invalid assignment")
            
    def __repr__(self):
        s = ""
        for row in self.data:
            s += str(row) + "\n"
        return s
    
# --------------------- New Functions --------------------- # 
    
    def shape(self):
        return (self.rows, self.cols)
    
    # Transposing flips rows and columns
    def transpose(self):
        new_data = []
        
        for j in range(self.cols):
            new_row = []
            for i in range(self.rows):
                new_row.append(self.data[i][j])
            new_data.append(new_row)
            
        return matrix(new_data)
    
    # Returning nth row of matrix M as new shaped matrix object
    def row(self, n):
        new_row = []
        
        for j in range(self.cols):
            new_row.append(self.data[n][j])
            
        return matrix([new_row])

    # Returning nth column of matrix M as new shaped matrix object
    def column(self, n):
        new_data = []
        
        for i in range(self.rows):
            new_data.append([self.data[i][n]])
            
        return matrix(new_data)
    
    # Returning matrix as list of lists
    def to_list(self):
        new_data = []
        
        for row in self.data:
            new_row = []
            for value in row:
                new_row.append(value)
            new_data.append(new_row)
            
        return new_data
    
    # Getting smaller matrix
    def block(self, n_0, n_1, m_0, m_1):
        new_data = []
        
        for i in range(n_0, n_1 + 1):
            new_row = []
            for j in range(m_0, m_1 + 1):
                new_row.append(self.data[i][j])
            new_data.append(new_row)
            
        return matrix(new_data)
```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python
# STANDALONE FUNCTIONS!!!
# Note: floats

def constant(n, m, c):
    data = []
    
    for i in range(n):
        row = []
        for j in range(m):
            row.append(float(c))
        data.append(row)
        
    return matrix(data)

def zeros(n, m):
    return constant(n, m, 0.0)

def ones(n, m):
    return constant(n, m, 1.0)

def eye(n):
    data = []
    
    for i in range(n):
        row = []
        for j in range(n):
            if i == j:
                row.append(1.0)
            else:
                row.append(0.0)
        data.append(row)
        
    return matrix(data)
```

```python
class matrix:

    def __init__(self, data, m=None):

        if isinstance(data, int) and isinstance(m, int):
            n = data
            self.rows = n
            self.cols = m
            
            self.data = []
            for i in range(n):
                row = []
                for j in range(m):
                    row.append(0)
                self.data.append(row)
        
        elif isinstance(data, list):
            
            for row in data:
                if not isinstance(row, list):
                    raise TypeError("Matrix must be a list of lists")

            first_length = len(data[0])
            for row in data:
                if len(row) != first_length:
                    raise ValueError("Rows must have the same number of columns")

            self.rows = len(data)
            self.cols = first_length

            self.data = []
            for row in data:
                new_row = []
                for value in row:
                    new_row.append(value)
                self.data.append(new_row)

        else:
            raise ValueError("Invalid initialization")

    def __getitem__(self, key):

        if isinstance(key, tuple):
            row_key, col_key = key

            if isinstance(row_key, int):
                row_key = slice(row_key, row_key + 1)
                
            if isinstance(col_key, int):
                col_key = slice(col_key, col_key + 1)
                
            new_data = []
            for i in range(*row_key.indices(self.rows)):
                new_row = []
                for j in range(*col_key.indices(self.cols)):
                    new_row.append(self.data[i][j])
                new_data.append(new_row)

            return matrix(new_data)
        
        else:
            return self.data[key]
        
    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.data[i][j] = value
        else:
            self.data[key] = value

    def assign(self, other):
        
        if isinstance(other, matrix):
            if self.rows != other.rows or self.cols != other.cols:
                raise ValueError("Matrices must be the same size")

            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other.data[i][j]

        elif isinstance(other, list):

            if len(other) != self.rows:
                raise ValueError("Incorrect number of rows")

            for row in other:
                if len(row) != self.cols:
                    raise ValueError("Incorrect number of columns")

            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other[i][j]

        else:
            raise TypeError("Invalid assignment")
            
    def __repr__(self):
        s = ""
        for row in self.data:
            s += str(row) + "\n"
        return s
    
    def shape(self):
        return (self.rows, self.cols)
    
    def transpose(self):
        new_data = []
        
        for j in range(self.cols):
            new_row = []
            for i in range(self.rows):
                new_row.append(self.data[i][j])
            new_data.append(new_row)
            
        return matrix(new_data)
    
    def row(self, n):
        new_row = []
        
        for j in range(self.cols):
            new_row.append(self.data[n][j])
            
        return matrix([new_row])

    def column(self, n):
        new_data = []
        
        for i in range(self.rows):
            new_data.append([self.data[i][n]])
            
        return matrix(new_data)
    
    def to_list(self):
        new_data = []
        
        for row in self.data:
            new_row = []
            for value in row:
                new_row.append(value)
            new_data.append(new_row)
            
        return new_data
    
    def block(self, n_0, n_1, m_0, m_1):
        new_data = []
        
        for i in range(n_0, n_1 + 1):
            new_row = []
            for j in range(m_0, m_1 + 1):
                new_row.append(self.data[i][j])
            new_data.append(new_row)
            
        return matrix(new_data)
        
    def scalarmul(self, c):
        new_data = []
        
        for i in range(self.rows):
            new_row = []
            for j in range(self.cols):
                new_row.append(self.data[i][j] * c)
            new_data.append(new_row)

        return matrix(new_data)
    
    def add(self, other): 
        new_data = []
        
        if self.rows != other.rows or self.cols != other.cols:
            raise ValueError("Matrices must have same dimensions for addition")

        for i in range(self.rows):
            new_row = []
            for j in range(self.cols):
                new_row.append(self.data[i][j] + other.data[i][j])
            new_data.append(new_row)

        return matrix(new_data)
    
    def sub(self, other):
        new_data = []
        
        if self.rows != other.rows or self.cols != other.cols:
            raise ValueError("Matrices must have same dimensions for subtraction")

        for i in range(self.rows):
            new_row = []
            for j in range(self.cols):
                new_row.append(self.data[i][j] - other.data[i][j])
            new_data.append(new_row)

        return matrix(new_data)
    
    def mat_mult(self, other):
        new_data = []

        if self.cols != other.rows:
            raise ValueError("Dimensions of matrices must be compatible for multiplication")

        for i in range(self.rows):
            new_row = []
            for j in range(other.cols):
                total = 0
                for k in range(self.cols):
                    total += self.data[i][k] * other.data[k][j]

                new_row.append(total)
            new_data.append(new_row)

        return matrix(new_data)
    
    def element_mult(self, other):
        new_data = []
    
        if self.rows != other.rows or self.cols != other.cols:
            raise ValueError("Matrices must have same dimensions for element-multiplication")

        for i in range(self.rows):
            new_row = []

            for j in range(self.cols):
                new_row.append(self.data[i][j] * other.data[i][j])
            new_data.append(new_row)

        return matrix(new_data)

    def equals(self, other):

        if self.rows != other.rows or self.cols != other.cols:
            return False

        for i in range(self.rows):
            for j in range(self.cols):
                if self.data[i][j] != other.data[i][j]:
                    return False

        return True
    
# --------------------- New Functions --------------------- #

    # 2*M; __rmul__
    # M*2; __mul__
    # M+N; __add__
    # M-N; __sub__
    # M*N; __mul__
    # M==N; __eq__

    # For M+N, python will call M.__add__(N)
    def __add__(self, other):
        return self.add(other)
    
    # For M-N, python will call M.__sub__(N)    
    def __sub__(self, other):
        return self.sub(other)
    
    # Has to handle scalar and matrix multiplication
    def __mul__(self, other):
        
        # Matrix * Matrix
        if isinstance(other, matrix):
            return self.mat_mult(other)
        
        # Matrix * Scalar
        elif isinstance(other, (int,float)):
            return self.scalarmul(other)
    
        else:
            raise TypeError("Unsupported multiplication")
    
    # Reverse multiplication (2*M)
    def __rmul__(self, other):
        
        if isinstance(other, (int, float)):
            return self.scalarmul(other)
        
        else:
            raise TypeError("Unsupported multiplication")
    
    # For M==N, python with call M.__eq__(N)
    def __eq__(self, other):
        return self.equals(other)
```
